```python
title_list =[]
rating_list = []
price_list = []

for pg_num in range(1,51):
  req = requests.get(f'https://books.toscrape.com/catalogue/page-{pg_num}.html')
  soup_pg = bs4.BeautifulSoup(req.text)

  for item in soup_pg.select('article h3 a'):
    title_list.append(item['title'])

  for n in range(0,60,3):
     rating_list.append(soup_pg.select('article p')[n]['class'][1])

  for n in range(1,60,3):
     price_list.append(soup_pg.select('article p')[n].string.replace('Â£',''))
```

```python
dic = dict(zip(title_list,zip(rating_list, price_list)))
df = pd.DataFrame.from_dict(dic, orient='index')
df
```

```python


import pandas as pd
import requests
import bs4
import time

all_books = []

for pg_num in range(1, 51):

    req = requests.get(
        f'https://books.toscrape.com/catalogue/page-{pg_num}.html'
    )

    soup_pg = bs4.BeautifulSoup(req.text, 'lxml')

    book_info_list = soup_pg.select('article h3 a')

    for book in book_info_list:

        try:
            # complete book url
            book_url = (
                'https://books.toscrape.com/catalogue/' +
                book['href']
            )

            res = requests.get(book_url, timeout=5)

            soup = bs4.BeautifulSoup(res.text, 'lxml')

            # column names
            column_names = [
                item.text for item in soup.select('tr th')
            ]

            # column values
            column_values = [
                item.text for item in soup.select('tr td')
            ]

            # dictionary
            d = dict(zip(column_names, column_values))

            # append dictionary
            all_books.append(d)

            print("Scraped:", book_url)

            # small delay
            time.sleep(0.1)

        except Exception as e:
            print("Error:", e)

# create dataframe once
new_df = pd.DataFrame(all_books)

print(new_df.head())

```
